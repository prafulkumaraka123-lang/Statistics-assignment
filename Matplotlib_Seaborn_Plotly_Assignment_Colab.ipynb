## Question 1 – What is Matplotlib, and why is it used for data visualization in Python? Explain the difference between `plt.plot()` and `plt.scatter()` with examples.

### Answer

**Matplotlib** is a widely used Python library for creating data visualizations such as line charts, bar charts, histograms, pie charts and scatter plots. It helps users understand patterns, trends, distributions and relationships in data.

**Why Matplotlib is used:**
- It supports many types of plots.
- It provides control over titles, labels, colors, axes, legends and styles.
- It works well with NumPy and Pandas.
- It can create static, animated and interactive visualizations.
- It is useful for exploratory data analysis and presenting results.

**Difference between `plt.plot()` and `plt.scatter()`:**
- `plt.plot()` is commonly used to draw lines connecting data points. It is suitable for showing trends, especially when x-values have an order, such as time.
- `plt.scatter()` displays individual points without connecting them by default. It is suitable for examining relationships, clusters and outliers between two numerical variables.

### Example program

````python
x = [1,  2,  3,  4,  5]
y = [2,  5,  4,  8,  7]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y, marker="o")
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.scatter(x, y, color="red", s=70)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

plt.tight_layout()
plt.show()
````

## Question 2 – What is a Figure and an Axes in Matplotlib? Explain the difference between `plt.subplot()` and `plt.subplots()` with an example.

### Answer

- **Figure:** The complete drawing area or canvas that contains one or more plots. It can contain titles, legends and multiple Axes.
- **Axes:** The individual plotting area inside a Figure where data is drawn. An Axes usually contains x- and y-axes, labels, ticks and plotted data.

**Difference:**
- `plt.subplot()` creates or selects a single subplot in a grid using a three-digit specification (rows, columns, plot position), or separate arguments.
- `plt.subplots()` creates a Figure and one or more Axes objects together. It is convenient when creating multiple plots and gives direct references to the Axes.

### Example using both

````python
# Example 1: plt.subplot()
plt.figure(figsize=(9, 3))
plt.subplot(1, 2, 1)
plt.plot([1, 2, 3], [1, 4, 9])
plt.title("Using plt.subplot()")

plt.subplot(1, 2, 2)
plt.bar(["A", "B", "C"], [3, 6, 4])
plt.title("Second subplot")
plt.tight_layout()
plt.show()

# Example 2: plt.subplots()
fig, axes = plt.subplots(1, 2, figsize=(9, 3))
axes[0].plot([1, 2, 3], [1, 4, 9], color="green")
axes[0].set_title("Line chart")

axes[1].bar(["A", "B", "C"], [3, 6, 4], color="orange")
axes[1].set_title("Bar chart")

fig.suptitle("Using plt.subplots()")
fig.tight_layout()
plt.show()
````

## Question 3 – Explain how to customize a line plot using color, linestyle, linewidth, marker and label. Write a program to plot a sine wave.

### Answer

Matplotlib provides several parameters to customize a line plot:

- **`color`:** Sets the color of the line, for example `"blue"` or `"red"`.
- **`linestyle`:** Sets the line pattern, such as `"-"` (solid), `"--"` (dashed), `":"` (dotted) or `"-."` (dash-dot).
- **`linewidth`:** Controls the thickness of the line.
- **`marker`:** Adds a symbol at each data point, such as `"o"`, `"s"` or `"^"`.
- **`label`:** Assigns a name to the plotted line, which can be displayed using `plt.legend()`.

### Python program – Sine wave

````python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.figure(figsize=(9, 5))
plt.plot(
    x, y,
    color="blue",
    linestyle="--",
    linewidth=2.5,
    marker="o",
    markevery=10,
    label="Sine Wave"
)

plt.title("Customized Sine Wave")
plt.xlabel("Angle (radians)")
plt.ylabel("sin(x)")
plt.legend()
plt.grid(True)
plt.show()
````

## Question 4 – What is Seaborn, and how is it different from Matplotlib? Explain `hue` with an example.

### Answer

**Seaborn** is a Python data visualization library built on top of Matplotlib. It provides a high-level interface for creating attractive statistical graphics and works especially well with Pandas DataFrames.

**Seaborn vs Matplotlib:**
- Matplotlib is a general-purpose plotting library that provides detailed, low-level control over plot elements.
- Seaborn provides higher-level functions for statistical visualization, with attractive default styles and convenient integration with DataFrames.
- Seaborn can simplify plots such as distributions, categorical comparisons and statistical relationships.
- Seaborn uses Matplotlib underneath, so Matplotlib functions can still be used to customize Seaborn plots.

**Meaning of `hue`:** The `hue` parameter maps a categorical variable to different colors. It helps distinguish groups within the same plot, such as different species or departments.

### Example

````python
df_hue = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 2, 3, 4, 5],
    "Marks": [35, 45, 55, 65, 75, 40, 52, 68, 80],
    "Group": ["A", "A", "A", "A", "A", "B", "B", "B", "B"]
})

sns.scatterplot(
    data=df_hue,
    x="Study_Hours",
    y="Marks",
    hue="Group",
    s=100
)
plt.title("Scatter Plot Using hue")
plt.show()
````

## Question 5 – Explain the difference between `plt.bar()` and `plt.barh()`. Create both charts using the given data.

Given data:

```python
categories = ["A", "B", "C", "D"]
values = [23, 17, 35, 29]
```

### Answer

- **`plt.bar()`** creates a vertical bar chart. Categories appear on the x-axis and values on the y-axis.
- **`plt.barh()`** creates a horizontal bar chart. Categories appear on the y-axis and values on the x-axis.

Vertical bars are useful for comparing categories when labels fit along the horizontal axis. Horizontal bars are useful when category labels are long or when a horizontal layout is easier to read.

### Python program

````python
categories = ["A", "B", "C", "D"]
values = [23, 17, 35, 29]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].bar(categories, values, color="skyblue")
axes[0].set_title("Vertical Bar Chart")
axes[0].set_xlabel("Categories")
axes[0].set_ylabel("Values")

axes[1].barh(categories, values, color="lightgreen")
axes[1].set_title("Horizontal Bar Chart")
axes[1].set_xlabel("Values")
axes[1].set_ylabel("Categories")

fig.tight_layout()
plt.show()
````

## Question 6 – What is a histogram, and what does its shape tell us about a dataset? Generate 1000 random values and plot a histogram with 30 bins.

### Answer

A **histogram** is a graph that represents the frequency distribution of numerical data. It divides the data range into intervals called **bins** and displays the number of observations in each interval using adjacent bars.

The shape of a histogram can reveal:
- **Center:** The approximate value around which observations are concentrated.
- **Spread:** How widely the data values are distributed.
- **Skewness:** Whether the distribution has a longer tail on the left or right.
- **Modality:** Whether the distribution has one peak or multiple peaks.
- **Unusual values:** Possible gaps or extreme observations.

A symmetric, bell-shaped histogram may indicate an approximately normal distribution. A histogram alone does not prove that data is normally distributed.

### Python program

````python
# Generate 1000 random values from a standard normal distribution
data = np.random.randn(1000)

plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, color="cornflowerblue", edgecolor="black")
plt.title("Histogram of 1000 Random Values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.3)
plt.show()
````

## Question 7 – Create a pie chart for the given fruit data, display percentages, set start angle to 120°, and explain `plt.axis("equal")`.

Given data:

```python
sizes = [40, 25, 20, 15]
labels = ["Apples", "Bananas", "Cherries", "Dates"]
```

### Answer

A **pie chart** represents data as slices of a circle. Each slice shows the proportion of the total represented by a category.

- `sizes` contains the values for each fruit.
- `labels` assigns a name to each slice.
- `autopct="%1.1f%%"` displays each slice's percentage to one decimal place.
- `startangle=120` rotates the first slice so that the pie chart starts at 120 degrees.

**Why use `plt.axis("equal")`?** It sets equal scaling on the x- and y-axes so the pie chart appears as a proper circle rather than an ellipse.

### Python program

````python
sizes = [40, 25, 20, 15]
labels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.figure(figsize=(7, 7))
plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    startangle=120
)
plt.title("Fruit Percentage Share")
plt.axis("equal")
plt.show()
````

## Question 8 – What is the purpose of `plt.annotate()`? Plot a sine wave and annotate its peak with an arrow and the label "Peak".

### Answer

`plt.annotate()` adds explanatory text to a plot and can point to a particular data location using an arrow. It is useful for highlighting important points, peaks, unusual observations or events.

Important parameters:
- `text`: The text to display.
- `xy`: The data coordinates of the point being highlighted.
- `xytext`: The position where the annotation text should appear.
- `arrowprops`: A dictionary used to customize the arrow.

### Python program

````python
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

peak_index = np.argmax(y)
peak_x = x[peak_index]
peak_y = y[peak_index]

plt.figure(figsize=(9, 5))
plt.plot(x, y, label="Sine Wave")
plt.annotate(
    "Peak",
    xy=(peak_x, peak_y),
    xytext=(peak_x + 0.7, peak_y - 0.25),
    arrowprops=dict(arrowstyle="->", color="red", linewidth=1.5),
    fontsize=12,
    color="red"
)

plt.title("Sine Wave with Peak Annotation")
plt.xlabel("Angle (radians)")
plt.ylabel("sin(x)")
plt.grid(True)
plt.legend()
plt.show()

print(f"Peak point: x = {peak_x:.3f}, y = {peak_y:.3f}")
````

## Question 9 – Use Seaborn to create a histogram and boxplot for the given scores. Identify any visible outliers.

Given scores:

```python
scores = [45, 67, 89, 23, 56, 78, 90, 34]
```

### Answer

A **histogram** groups numerical values into bins and shows their frequency. A **boxplot** summarizes a dataset using the median, lower and upper quartiles, and whiskers. Individual points beyond the whiskers may be shown as potential outliers.

The program:
1. Loads the scores into a Pandas DataFrame with the column name `Score`.
2. Uses `sns.histplot()` to create a histogram.
3. Uses `sns.boxplot()` to create a boxplot.
4. Checks for potential outliers using the 1.5 × IQR rule.

**Outlier calculation for this dataset:** Using the standard median-of-halves quartiles, Q1 = 39.5, Q3 = 83.5, and IQR = 44. The fences are -26.5 and 149.5. All the supplied scores lie within these fences, so there are **no outliers** by the 1.5 × IQR rule.

### Python program

````python
scores = [45, 67, 89, 23, 56, 78, 90, 34]
df_scores = pd.DataFrame({"Score": scores})

display(df_scores)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

sns.histplot(data=df_scores, x="Score", bins=5, kde=False, ax=axes[0])
axes[0].set_title("Histogram of Scores")

sns.boxplot(data=df_scores, y="Score", ax=axes[1])
axes[1].set_title("Boxplot of Scores")

plt.tight_layout()
plt.show()

# Identify potential outliers using the 1.5 x IQR rule
q1 = df_scores["Score"].quantile(0.25)
q3 = df_scores["Score"].quantile(0.75)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

outliers = df_scores[
    (df_scores["Score"] < lower_fence) |
    (df_scores["Score"] > upper_fence)
]["Score"].tolist()

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)
print("Potential outliers:", outliers if outliers else "No outliers")
````

## Question 10 – What is Plotly, and how is it different from Matplotlib and Seaborn? Create an interactive scatter plot using Plotly Express.

Given data:

```python
x = [1, 2, 3, 4, 5]
y = [10, 14, 22, 18, 27]
```

### Answer

**Plotly** is a Python visualization library used to create interactive charts and dashboards. Its charts can support hovering, zooming, panning, selecting and exporting.

**Plotly compared with Matplotlib and Seaborn:**
- **Matplotlib:** A general-purpose plotting library that provides detailed control and is widely used for static plots.
- **Seaborn:** A high-level statistical visualization library built on Matplotlib, with convenient functions and attractive default styles.
- **Plotly:** Focuses on interactive, web-based visualizations and is useful for dashboards and exploring data in a browser.

**One advantage of Plotly:** Users can hover over data points to see their values and zoom into parts of a chart without changing the original code. This makes interactive exploration easier than with a typical static plot.

### Python program using `plotly.express`

````python
x = [1, 2, 3, 4, 5]
y = [10, 14, 22, 18, 27]

df_plotly = pd.DataFrame({"x": x, "y": y})

fig = px.scatter(
    df_plotly,
    x="x",
    y="y",
    title="Interactive Scatter Plot",
    labels={"x": "X Values", "y": "Y Values"},
    hover_data=["x", "y"]
)

fig.update_traces(marker=dict(size=12))
fig.show()
````
